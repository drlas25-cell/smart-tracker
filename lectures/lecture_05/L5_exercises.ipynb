```python
class Student:
    def __init__(self, name, grades):
        # TODO: خزّن الاسم والدرجات من المعاملات — استبدل المؤقت
        self.name = name
        self.grades = grades

    def average(self):
        # TODO: لفّ على الدرجات ورجّع المتوسط (ناتج مؤقت)
        if not self.grades:
            return 0

        total =0
        for grade in self.grades:
            total += grade
        return total/len(self.grades)

s = Student("Sara", [85, 90, 76])
print(s.name)
print(s.average())
```

```python

class BankAccount:
    def __init__(self, owner, balance=0):
        # TODO: خزّن اسم المالك والرصيد
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        # TODO: زوّد الرصيد بمبلغ الإيداع
        self.balance += amount
        print(f"تم إيداع {amount}. الرصيد الحالي: {self.balance}")

    def withdraw(self, amount):
        # TODO: لو المبلغ أكبر من الرصيد اطبع رسالة واقف — غير كده اخصم
        if amount > self.balance:
            print("رصيد غير كافٍ")
        else:
            self.balance -= amount
            print(f"تم سحب {amount}. الرصيد المتبقي: {self.balance}")

acc = BankAccount("Omar", 100)
acc.deposit(50)
acc.withdraw(30)
acc.withdraw(200)





```

```python
class Student:
    def __init__(self, name, grades):
        # TODO: خزّن الاسم والدرجات من المعاملات — استبدل المؤقت
        self.name = name
        self.grades = grades

    def average(self):
        # TODO: لفّ على الدرجات ورجّع المتوسط (ناتج مؤقت)
        if not self.grades:
            return 0

        total =0
        for grade in self.grades:
            total += grade
        return total/len(self.grades)

s = Student("Sara", [85, 90, 76])
print(s.name)
print(s.average())





class Classroom:
    def __init__(self, room):
        # TODO: اسم القاعة + قائمة طلاب فاضية
        self.room = room
        self.students = []

    def add(self, student):
        # TODO: ضيف الطالب لقائمة الطلاب
        self.students.append(student)

    def best(self):
        # TODO: رجّع الطالب الأعلى بمتوسط الدرجات (لفّ على القائمة)
        if not self.students:
            return None

        best_student = self.students[0]
        for student in self.students:
            if student.average() > best_student.average():
                best_student = student
        return best_student

# تجربة الكود وتأكيد العمل
room = Classroom("A101")
room.add(Student("Sara", [85, 90, 76]))
room.add(Student("Ahmad", [90, 95, 88]))
room.add(Student("Noor", [70, 80, 75]))

top_student = room.best()
if top_student:
    print(f"أفضل طالب في القاعة {room.room} هو: {top_student.name} بمعدل {top_student.average():.2f}")
```
