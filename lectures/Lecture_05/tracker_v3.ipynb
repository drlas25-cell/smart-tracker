```python
%%writefile tracker/functions.py
# tracker/functions.py
import os
import csv

def add_record(records, record):
    """Append one record dict to the list."""
    # TODO: أضف السجل للقائمة وارجعها
    records.append(record)
    return records

def filter_records(records, category):
    """Return records matching the given category."""
    # TODO: لفّ واشرط — أعد قائمة جديدة بالمطابقين
    return [r for r in records if r.get('category') == category]

def summarize(records, key='amount_egp'):
#def summarize(records):
    """Return count, total, average, biggest, smallest of the specified key."""
    #"""Return count, total, average, biggest, smallest of amount_egp."""
    # TODO: نمط الملخص — لو القائمة فاضية رجّع أصفار
    amounts = [float(r[key]) for r in records if key in r and r[key] is not None]
    #amounts = [float(r['amount_egp']) for r in records if 'amount_egp' in r]

    if not amounts:
        return {
            'count': 0,
            'total': 0.0,
            'average': 0.0,
            'biggest': 0.0,
            'smallest': 0.0
        }

    total_val = sum(amounts)
    count_val = len(amounts)
    return {
        'count': count_val,
        'total': total_val,
        'average': total_val / count_val,
        'biggest': max(amounts),
        'smallest': min(amounts)
    }

def save_records(records, path):
    """Write records to CSV with a header row."""
    # TODO: csv.writer — صف العناوين أول حاجة
    if not records:
        return

    headers = list(records[0].keys())
    with open(path, 'w', newline='', encoding='utf-8') as f:
        writer = csv.DictWriter(f, fieldnames=headers)
        writer.writeheader()
        writer.writerows(records)


def load_records(path):
    """Read CSV back into dicts; a missing file returns an empty list."""
    # TODO: except FileNotFoundError — لو الملف مش موجود رجّع قائمة فاضية
    # TODO: amount_egp -> float بعد القراءة
    if not os.path.exists(path):
        return []

    records = []
    try:
        with open(path, 'r', newline='', encoding='utf-8') as f:
            reader = csv.DictReader(f)
            for row in reader:
                if 'amount_egp' in row:
                    try:
                        row['amount_egp'] = float(row['amount_egp'])
                    except ValueError:
                        pass
                records.append(dict(row))
    except FileNotFoundError:
        return []
    return records

def delete_record(records, index):
    """Delete the record at index and return it."""
    if index < 0 or index >= len(records):
        return None
    return records.pop(index)


def update_record(records, index, field, value):
    """Change one field of the record at index."""
    # نتأكد أن الفهرس صحيح وموجود داخل القائمة
    if index < 0 or index >= len(records):
        return None

    # نأخذ السجل المطلوب
    record = records[index]

    # نتحقق إذا كان الحقل موجوداً بالفعل في السجل
    if field in record:
        record[field] = value
        return record

    return None



```

```python
%%writefile tracker/models.py
# tracker/models.py
import csv
import os

from tracker.functions import (
    add_record, filter_records, summarize,
    save_records, load_records
)

class Record:
    """One tracker row: the five project fields."""
    #def __init__(self, date, category, amount_egp,payment_method, quantity):
        # TODO: خزّن الحقول الخمسة على الكائن — المبلغ يتحفظ رقمًا
        #pass


    def __init__(self, issue_id, title, priority, status, hours_spent, days_open, assigned_to, platform):
        self.issue_id = int(issue_id)
        self.title = str(title)
        self.priority = str(priority)
        self.status = str(status)
        self.hours_spent = float(hours_spent)
        self.days_open = int(days_open)
        self.assigned_to = str(assigned_to)
        self.platform = str(platform)




    def to_dict(self):
        # TODO: رجّع قاموس بالحقول الخمسة — بنفس الترتيب
        #pass
        return {
            "issue_id": self.issue_id,
            "title": self.title,
            "priority": self.priority,
            "status": self.status,
            "hours_spent": self.hours_spent,
            "days_open": self.days_open,
            "assigned_to": self.assigned_to,
            "platform": self.platform
        }

    def __str__(self):
        # TODO: رجّع سطرًا منسّقًا يظهر مباشرة في الشاشة
        #pass
        return f"ID:{self.issue_id} | Title:{self.title} | Priority: {self.priority} | Status: {self.status} | Hours_spent:{self.hours_spent} | Days_open: {self.days_open} | Assigned_to: {self.assigned_to} | Platform: {self.platform}"


class Tracker:
    """A list of Record objects + every behavior the menu needs."""

    def __init__(self):
         # TODO: self.records = قائمة فاضية
        #pass
        self.records = []

    def add(self, record):
         # TODO: ضيف الكائن لقائمة السجلات
        #pass
        self.records.append(record)

    def filter_by(self, category):
        # TODO: قائمة جديدة بسجلات التصنيف المطلوب فقط
        #pass
        return [r for r in self.records if r.status == category]
       # لتصفيه السجلات حسب حاله المشكله مثلا
       # return [r for r in self.records if r.status.strip().title() == categorystrip().title()]




    def summary(self):
        # TODO: قاموس الملخص — العدد والمجموع والمتوسط والأعلى والأدنى
        #       للمبلغ الأساسي — ولو القائمة فاضية رجّع أصفار
        hours = [r.hours_spent for r in self.records]
        if not hours:
            return {
                'count': 0,
                'total': 0.0,
                'average': 0.0,
                'biggest': 0.0,
                'smallest': 0.0
            }
        total_val = sum(hours)
        count_val = len(hours)
        return {
            'count': count_val,
            'total': total_val,
            'average': total_val / count_val,
            'biggest': max(hours),
            'smallest': min(hours)
        }

    def save(self, path):
        # TODO: csv + to_dict — صف العناوين أول حاجة وبعده السجلات
        # pass
        if not self.records:
            return
        headers = ["issue_id", "title", "priority", "status", "hours_spent", "days_open", "assigned_to", "platform"]
        with open(path, 'w', newline='', encoding='utf-8') as f:
            writer = csv.DictWriter(f, fieldnames=headers)
            writer.writeheader()
            for r in self.records:

                writer.writerow(r.to_dict())




    def load(self, path):
        # TODO: اقرأ الصفوف وحوّلها لسجلات الكائن داخل المتتبّع
        #       ملف مش موجود → سيب القائمة زي ما هي
        #       المبلغ لازم يرجع رقمًا بعد القراءة


        if not os.path.exists(path):
            return
        self.records = []
        try:
            with open(path, 'r', newline='', encoding='utf-8') as f:
                reader = csv.DictReader(f)
                for row in reader:
                    record_obj = Record(
                        issue_id=row["issue_id"],
                        title=row["title"],
                        priority=row["priority"],
                        status=row["status"],
                        hours_spent=row["hours_spent"],
                        days_open=row["days_open"],
                        assigned_to=row["assigned_to"],
                        platform=row["platform"]
                    )
                    self.add(record_obj)
        except FileNotFoundError:
            pass


        # tracker/models.py — الطريقة دي تنزل جوّا كلاس المتتبّع
    def top_n(self, field, n):
        """Return the n records with the biggest value of field."""
        # TODO: لفّ على السجلات وتتبّع أكبر قيمة — ورتّبهم تنازليًا
        #pass
        sorted_records = sorted(self.records, key=lambda r: getattr(r, field), reverse=True)
        return sorted_records[:n]






    # tracker/models.py — الطريقة دي تنزل جوّا كلاس المتتبّع
    def baseline_predict(self, record):
        """Rule-based prediction - the benchmark the ML must beat."""
        # نهج يعتمد على if/else البسيطة بناءً على حقل الأولوية (Priority)
        priority = str(record.get('priority', 'Medium')).lower()

        if priority == 'high':
            return 16.0  # المشاكل العالية تحتاج وقت أطول
        elif priority == 'medium':
            return 8.0   # المشاكل المتوسطة كقيمة افتراضية
        else:
            return 4.0   # المشاكل منخفضة الأولوية



```
