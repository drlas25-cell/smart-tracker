```python
import pandas as pd
from sklearn.linear_model import LinearRegression

study = pd.DataFrame({
    "hours": [1, 2, 3, 4, 5, 6, 7, 8],
    "score": [52, 55, 61, 63, 70, 72, 78, 81],
})
X = study[["hours"]]
y = study["score"]

model = LinearRegression().fit(X, y)
new_student = pd.DataFrame({"hours": [9]})   # طالب جديد - صف واحد بجدول
pred = float(model.predict(new_student)[0])
print(round(pred, 1))     # 85.7

assert 84 < pred < 88, "تنبؤ في حدود المعقول"
print("A1: ممتاز")
```

```python
import pandas as pd
from sklearn.linear_model import LinearRegression

orders = pd.DataFrame({
    "method": ["express", "standard", "express", "pickup", "standard",
               "pickup", "express", "standard"],
    "minutes": [25, 45, 30, 15, 50, 18, 22, 40],
    "fee": [12.0, 7.0, 13.0, 4.0, 6.0, 4.5, 14.0, 7.5],
})
X = pd.get_dummies(orders[["minutes", "method"]],
                   columns=["method"], dtype=float)
y = orders["fee"]
train_cols = list(X.columns)          # احفظ أسماء الأعمدة

model = LinearRegression().fit(X, y)
new_orders = pd.DataFrame([
    {"method": "express", "minutes": 20},
    {"method": "pickup", "minutes": 25},
    {"method": "standard", "minutes": 35},
    {"method": "express", "minutes": 40},
    {"method": "pickup", "minutes": 12},
])
X_new = pd.get_dummies(new_orders[["minutes", "method"]],
                       columns=["method"], dtype=float)
X_new = X_new.reindex(columns=train_cols, fill_value=0)

preds = model.predict(X_new)
print([round(float(p), 2) for p in preds])
# [13.63, 3.3, 7.95, 11.4, 4.75]

assert list(X_new.columns) == train_cols, "الأعمدة بنفس أسماء التدريب"
print("A2: ممتاز")
```

```python
import pandas as pd
from sklearn.tree import DecisionTreeRegressor

cars = pd.DataFrame({
    "year": [2015, 2018, 2012, 2020, 2016, 2019, 2013, 2017, 2021, 2014],
    "km": [120000, 80000, 150000, 40000, 110000, 60000,
           140000, 90000, 30000, 130000],
    "price": [180, 260, 130, 330, 200, 290, 140, 240, 350, 150],
})
X = cars[["year", "km"]]
y = cars["price"]
tree = DecisionTreeRegressor(max_depth=3, random_state=0).fit(X, y)

new_cars = pd.DataFrame({"year": [2019, 2016, 2022],
                         "km": [55000, 105000, 25000]})
broken = new_cars[["km"]]            # year ناقصة من الجدول
try:
    tree.predict(broken)
    print("غلط: الموديل المفروض يقع هنا")
except ValueError as e:
    print("الخطأ كما هو:", str(e)[:60])
    # Feature names seen at fit time, yet now missing: - year

fixed = broken.reindex(columns=["year", "km"], fill_value=0)
print([round(float(p)) for p in tree.predict(fixed)])   # [250, 150, 250]

assert list(fixed.columns) == ["year", "km"], "العمود الناقص رجع بصفر"
print("A3: ممتاز")
```

```python
import pandas as pd
from sklearn.linear_model import LinearRegression

gym = pd.DataFrame({
    "minutes": [10, 20, 30, 40, 50, 60, 70, 80],
    "calories": [52, 105, 161, 208, 250, 310, 355, 405],
})
X = gym[["minutes"]]
y = gym["calories"]
model = LinearRegression().fit(X, y)


def predict_calories(record):
    """تنبؤ من قاموس واحد - صف واحد بجدول"""
    one = pd.DataFrame([record])
    return round(float(model.predict(one)[0]), 1)


print(predict_calories({"minutes": 45}))   # 230.8
print(predict_calories({"minutes": 90}))   # 456.3

assert predict_calories({"minutes": 45}) == 230.8, "القيمة 45 دقيقة"
print("A4: ممتاز")
```
