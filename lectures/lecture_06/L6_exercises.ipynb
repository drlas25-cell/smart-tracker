### A4 — سير عمل تعلم الآلة لمشروع (Issue Tracker System)

1. **Define the problem:** التنبؤ بعدد الساعات اللازمة لحل المشكلة (Predicting `hours_spent`) بناءً على الأولوية ونوع المنصة والمسؤول لتقدير الجهد بدقة.
2. **Get data:** تجميع سجلات التذاكر المغلقة سابقاً من قاعدة بيانات نظام تتبع المشاكل الحالي (Issue Tracker Database).
3. **Clean:** التعامل مع التذاكر التي تحتوي على قيم مفقودة في حقل المسؤول `assigned_to` أو أرقام سالبة غير منطقية في حقل الأيام المفتوحة `days_open`.
4. **Explore:** دراسة العلاقة بين أولوية المشكلة `priority` وعدد الساعات المستغرقة `hours_spent` لمعرفة هل القضايا العاجلة تُحل أسرع أم تأخذ وقتاً أطول لتعقيدها.
5. **Train:** تدريب نموذج انحدار (Regression Model)  باستخدام بيانات التذاكر التاريخية للتنبؤ بقيمة `hours_spent` المستمرة.
6. **Evaluate:** قياس دقة النموذج باستخدام متوسط الخطأ المطلق (MAE) لمقارنة الساعات المتوقعة بالساعات الفعلية المستغرقة للتأكد من كفاءة التنبؤ.
7. **Predict:** دمج النموذج داخل النظام لتوليد تقدير تلقائي للوقت المتوقع فور إنشاء تذكرة جديدة ومساعدة الإدارة في توزيع المهام بشكل أفضل.
