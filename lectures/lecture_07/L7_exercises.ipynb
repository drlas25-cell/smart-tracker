```python
import numpy as np

x = np.array([12.0, 18.0, 9.0, 21.0, 15.0])
norm = (x - x.mean()) / x.std()                  # TODO: عدّل لو غلط
print(round(norm.mean(), 6))                     # 0.0
print(round(norm.std(), 6))                      # 1.0

m = np.array([[1, 2, 3],
              [4, 5, 6]])
print(m.shape)                                   # (2, 3)
print(m.mean(axis=0))                            # [2.5 3.5 4.5]
print(m.mean(axis=1))                            # [2. 5.]

assert abs(norm.mean()) < 1e-9, "المتوسط ~ صفر"
assert abs(norm.std() - 1.0) < 1e-9, "الانحراف = 1"
assert m.sum() == 21 and m.shape == (2, 3)
print("A1: ممتاز")
```

```python
import os
import pandas as pd

os.makedirs("data", exist_ok=True)
coffee = [
    {"day": "Mon", "drink": "espresso", "cups": 12, "revenue": 36.0},
    {"day": "Mon", "drink": "latte",    "cups": 18, "revenue": 90.0},
    {"day": "Tue", "drink": "espresso", "cups": 10, "revenue": 30.0},
    {"day": "Tue", "drink": "latte",    "cups": 22, "revenue": 110.0},
    {"day": "Wed", "drink": "espresso", "cups": 14, "revenue": 42.0},
    {"day": "Wed", "drink": "tea",      "cups": 9,  "revenue": 27.0},
    {"day": "Thu", "drink": "latte",    "cups": 20, "revenue": 100.0},
    {"day": "Fri", "drink": "espresso", "cups": 16, "revenue": 48.0},
    {"day": "Fri", "drink": "latte",    "cups": 25, "revenue": 125.0},
    {"day": "Sat", "drink": "latte",    "cups": 30, "revenue": 150.0},
    {"day": "Sat", "drink": "tea",      "cups": 12, "revenue": 36.0},
]
pd.DataFrame(coffee).to_csv("data/coffee_demo.csv", index=False)

df = pd.read_csv("data/coffee_demo.csv")
print(df.shape)                                  # (11, 4)
print(df["drink"].nunique())                     # 3
print(round(df["cups"].mean(), 1))               # 17.1
print(df["revenue"].sum())                       # 794.0
best_day = df.groupby("day")["revenue"].sum().idxmax()
print(best_day)                                  # Sat

assert df.shape == (11, 4), "11 صفوف 4 أعمدة"
assert df["drink"].nunique() == 3, "3 مشروبات"
assert round(df["cups"].mean(), 1) == 17.1, "متوسط الأكواب"
assert df["revenue"].sum() == 794.0, "إجمالي الإيراد"
assert best_day == "Sat", "أعلى يوم"
print("A2: ممتاز")
```

```python
fri = df[df["day"] == "Fri"].sort_values("revenue", ascending=False)
print(fri["drink"].iloc[0])                      # latte
assert list(fri["revenue"]) == [125.0, 48.0], "إيراد الجمعة"
top = df.sort_values("revenue", ascending=False).head(1)
assert top["day"].iloc[0] == "Sat", "أعلى صف السبت"
print("A3: ممتاز")
```

```python
by_day = df.groupby("day")["revenue"].sum().sort_values(ascending=False)
by_drink = df.groupby("drink")["cups"].mean()
print(by_day.idxmax(), by_day.max())             # Sat 186.0
print(round(by_drink["latte"], 1))               # 23.0
assert by_day.idxmax() == "Sat" and by_day.max() == 186.0
assert round(by_drink["latte"], 1) == 23.0, "اللاتيه"
assert round(by_drink["espresso"], 1) == 13.0, "الإسبريسو"
print("A4: ممتاز")
```

```python
%%writefile tracker/analysis.py
import csv
import os

"""Dataset analysis - load, clean, and summarize the project dataset."""
import pandas as pd

CSV_PATH = "data/sample_records_250.csv"


#def load_dataset(path=CSV_PATH):
    # TODO: اقرا الملف وارجع النتيجة
#    pass


#def clean_dataset(df):
    # TODO: drop_duplicates ثم اتعامل مع النواقص
    # (احذف النواقص أو عوّضها — واكتب قرارك في التقرير)
 #   pass


#def summary_by_category(df, category_col, value_col):
    # TODO: groupby + mean
#    pass



def load_dataset(path=CSV_PATH):
    """Reads the CSV file into a pandas DataFrame."""
    if not os.path.exists(path):
        print(f"Error: File not found at {path}")
        return pd.DataFrame()
    return pd.read_csv(path)




def clean_dataset(df):
    """Removes duplicates and handles missing values."""
    if df.empty:
        return df

    # إزالة الصفوف المكررة بالكامل
    df_cleaned = df.drop_duplicates()

    # التعامل مع القيم الناقصة (ملء القيم الفارغة في الحقول النصية بـ 'Unknown' والرقمية بـ المتوسط)
    for col in df_cleaned.columns:
        if df_cleaned[col].isnull().any():
            if df_cleaned[col].dtype in ['float64', 'int64']:
                df_cleaned[col] = df_cleaned[col].fillna(df_cleaned[col].mean())
            else:
                df_cleaned[col] = df_cleaned[col].fillna('Unknown')

    return df_cleaned


def summary_by_category(df, category_col, value_col):
    """Groups by category_col and calculates the mean of value_col."""
    if df.empty or category_col not in df.columns or value_col not in df.columns:
        return pd.DataFrame()

    # تجميع البيانات وحساب المتوسط لكل تصنيف
    summary_df = df.groupby(category_col)[value_col].mean().reset_index()
    return summary_df


def count_by_category(df, category_col):
    """Groups by category_col and calculates the count of occurrences."""
    if df.empty or category_col not in df.columns:
        return pd.DataFrame()

    # تجميع البيانات وحساب العدد لكل تصنيف
    count_df = df.groupby(category_col).size().reset_index(name='issue_count')
    return count_df





df_raw = load_dataset()
print(f"Raw dataset shape: {df_raw.shape}")



df_clean = clean_dataset(df_raw)
print(f"Cleaned dataset shape: {df_clean.shape}")



cleaned_file_path = "data/cleaned_sample_records_250.csv"
df_clean.to_csv(cleaned_file_path, index=False)
print(f"تم حفظ الملف بنجاح في المسار: {cleaned_file_path}")



# حساب متوسط الساعات المستغرقة (hours_spent) لكل مستوى أولوية (priority)
summary_priority = summary_by_category(df_clean, "priority", "hours_spent")
display(summary_priority)


#   حساب متوسط الساعات المستغرقة لكل حالة (Status)
summary_status_hours = summary_by_category(df_clean, "status", "hours_spent")
print("--- متوسط الساعات المستغرقة لكل حالة (Status) ---")
display(summary_status_hours)

#  حساب متوسط الساعات المستغرقة لكل مطور (Assigned To)
summary_dev_hours = summary_by_category(df_clean, "assigned_to", "hours_spent")
print("\n--- متوسط الساعات المستغرقة لكل مطور (Assigned To) ---")
display(summary_dev_hours)

#  حساب متوسط الساعات المستغرقة لكل منصة (Platform)
summary_platform_hours = summary_by_category(df_clean, "platform", "hours_spent")
print("\n--- متوسط الساعات المستغرقة لكل منصة (Platform) ---")
display(summary_platform_hours)


#   حساب عدد المشاكل في كل حالة (Status)
status_counts_func = count_by_category(df_clean, "status")
print("--- عدد المشاكل في كل حالة (Status) باستخدام الدالة الجديدة ---")
display(status_counts_func)

#   حساب عدد المشاكل في كل منصة (Platform)
platform_counts_func = count_by_category(df_clean, "platform")
print("\n--- عدد المشاكل في كل منصة (Platform) باستخدام الدالة الجديدة ---")
display(platform_counts_func)

#   حساب عدد المشاكل في كل مطور (Platform)
platform_counts_func = count_by_category(df_clean, "assigned_to")
print("\n--- عدد المشاكل لكل مطور (assigned_to) باستخدام الدالة الجديدة ---")
display(platform_counts_func)


```
