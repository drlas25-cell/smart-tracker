```python
def average(values):
    # TODO: رجّع متوسط القائمة
    return sum(values) / len(values)
    pass
print(average([82, 90, 76]) )
```

```python
def min_max(values):
    # TODO: لفّ على القائمة وتتبّع أكبر وأصغر — ورجّع (الأصغر, الأكبر)
    biggest = values[0]
    smallest = values[0]
    for v in values:
     if v > biggest:
       biggest = v
     if v < smallest:
       smallest = v
    return smallest, biggest
    pass

low, high = min_max([82,90,76])
print(low,high)

low, high = min_max([5])
print(low, high)

low, high = min_max([-3, -9, 2])
print(low, high)
```

```python
lines = ["date,category", "2026-10-01,food", "2026-10-02,drink"]
path = "my_lines.txt"
content = ""

# TODO: with للكتابة — كل سطر في سطر جديد
# TODO: with للقراءة — النتيجة كلها في المتغيّر اللي فوق
# TODO: except FileNotFoundError لو الملف مش موجود

# كتابة السطور في الملف
try:
    with open(path, 'w', encoding='utf-8') as f:
        for line in lines:
            f.write(line + '\n')
except Exception as e:
    print(f"حدث خطأ أثناء الكتابة: {e}")

# قراءة الملف
try:
    with open(path, 'r', encoding='utf-8') as f:
        content = f.read()
except FileNotFoundError:
    print(f"خطأ: الملف '{path}' غير موجود!")
except Exception as e:
    print(f"حدث خطأ غير متوقع: {e}")

print("محتوى الملف:")
print(content)


# except FileNotFoundError لو الملف مش موجود
print("\n--- تجربة قراءة ملف غير موجود ---")
NoFile_path = "NoFile.txt"
try:
    with open(NoFile_path, 'r', encoding='utf-8') as f:
        Nocontent = f.read()
except FileNotFoundError:
    print(f": الملف '{NoFile_path}' غير موجود!")
```
