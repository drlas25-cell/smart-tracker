```python
%%writefile tracker/functions.py
# tracker/functions.py
import os
import csv

def add_record(records, record):
    """Append one record dict to the list."""
    # TODO: أضف السجل للقائمة وارجعها
    records.append(record)
    return records

def filter_records(records, category):
    """Return records matching the given category."""
    # TODO: لفّ واشرط — أعد قائمة جديدة بالمطابقين
    return [r for r in records if r.get('category') == category]





def summarize(records, key='amount_egp'):
#def summarize(records):
    """Return count, total, average, biggest, smallest of the specified key."""
    #"""Return count, total, average, biggest, smallest of amount_egp."""
    # TODO: نمط الملخص — لو القائمة فاضية رجّع أصفار
    amounts = [float(r[key]) for r in records if key in r and r[key] is not None]
    #amounts = [float(r['amount_egp']) for r in records if 'amount_egp' in r]

    if not amounts:
        return {
            'count': 0,
            'total': 0.0,
            'average': 0.0,
            'biggest': 0.0,
            'smallest': 0.0
        }

    total_val = sum(amounts)
    count_val = len(amounts)
    return {
        'count': count_val,
        'total': total_val,
        'average': total_val / count_val,
        'biggest': max(amounts),
        'smallest': min(amounts)
    }




def save_records(records, path):
    """Write records to CSV with a header row."""
    # TODO: csv.writer — صف العناوين أول حاجة
    if not records:
        return

    headers = list(records[0].keys())
    with open(path, 'w', newline='', encoding='utf-8') as f:
        writer = csv.DictWriter(f, fieldnames=headers)
        writer.writeheader()
        writer.writerows(records)


def load_records(path):
    """Read CSV back into dicts; a missing file returns an empty list."""
    # TODO: except FileNotFoundError — لو الملف مش موجود رجّع قائمة فاضية
    # TODO: amount_egp -> float بعد القراءة
    if not os.path.exists(path):
        return []

    records = []
    try:
        with open(path, 'r', newline='', encoding='utf-8') as f:
            reader = csv.DictReader(f)
            for row in reader:
                if 'amount_egp' in row:
                    try:
                        row['amount_egp'] = float(row['amount_egp'])
                    except ValueError:
                        pass
                records.append(dict(row))
    except FileNotFoundError:
        return []
    return records


```

```python
def delete_record(records, index):
    """Delete the record at index and return it."""
    if index < 0 or index >= len(records):
        return None
    return records.pop(index)


def update_record(records, index, field, value):
    """Change one field of the record at index."""
    # نتأكد أن الفهرس صحيح وموجود داخل القائمة
    if index < 0 or index >= len(records):
        return None

    # نأخذ السجل المطلوب
    record = records[index]

    # نتحقق إذا كان الحقل موجوداً بالفعل في السجل
    if field in record:
        record[field] = value
        return record

    return None
```
