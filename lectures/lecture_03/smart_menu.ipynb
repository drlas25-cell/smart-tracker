```python
while True:
    print("=" * 40 )
    print("1) ﻋﺮض اﻟﻜﻞ")
    print("2) إﺿﺎﻓﺔ ﺳﺠﻞ")
    print("3) ﺗﺼﻔﯿﺔ ﺣﺴﺐ اﻟﺘﺼﻨﯿﻒ")
    print("4) ﻣﻠﺨﺺ")
    print("5) ﺑﺤﺚ ﺑﺎﻟﻜﻠﻤﺔ")
    print("6) ﺧﺮوج")
    print("=" * 40 )

    choice = input("اﺧﺘﺎر: ")
    print("=" * 40 )

    if choice == "1":
        print("1) ﻋﺮض اﻟﻜﻞ")
        for r in issues_dataset:
           # print(r)
            print(f"ID: {r['issue_id']} | Title: {r['title']} | Priority: {r['priority']} | Status: {r['status']} | Hours: {r['hours_spent']} | Days Open: {r['days_open']} | Assigned To: {r['assigned_to']} | Platform: {r['platform']}")


    elif choice == '2':
        print("2) إﺿﺎﻓﺔ ﺳﺠﻞ")
        try:
            issue_id = int(input("Enter issue ID (integer): "))
            title = input("Enter issue title: ").strip()
            priority = input("Enter priority (Low/Medium/High/Critical): ").strip().title()
            status = input("Enter status (Open/In Progress/Closed): ").strip().title()
            hours_spent = float(input("Enter hours spent (decimal/number): "))
            days_open = int(input("Enter days open (integer): "))
            assigned_to = input("Enter assignee name: ").strip().title()
            platform = input("Enter platform: ").strip().title()

            new_record = {
                "issue_id": issue_id,
                "title": title,
                "priority": priority,
                "status": status,
                "hours_spent": hours_spent,
                "days_open": days_open,
                "assigned_to": assigned_to,
                "platform": platform
            }
            issues_dataset.append(new_record)
            print("\nRecord added successfully!")
        except ValueError:
            print("\nError: Please make sure to enter correct numeric values for ID, hours, and days.")
        pass

    elif choice == '3':
        print("3) ﺗﺼﻔﯿﺔ ﺣﺴﺐ اﻟﺘﺼﻨﯿﻒ")
        target_Status = input("Enter status name to filter (e.g. Open, In Progress, Closed): ").strip().lower()
        found = False
        for r in issues_dataset:
            if r['status'].lower() == target_Status:
                print(f"ID: {r['issue_id']} | Title: {r['title']} | Status: {r['status']}")
                found = True
       # if not found:
       #    print("No records found matching this Status.")
        pass


    elif choice == '4':
        print("4) ﻣﻠﺨﺺ")

        if len(issues_dataset) == 0:
            print("No records found.")

        else:
            total_hours = 0.0
            max_hours = float('-inf')
            min_hours = float('inf')
            count = 0

            for r in issues_dataset:
                val = r['hours_spent']
                total_hours += val
                if val > max_hours:
                    max_hours = val
                if val < min_hours:
                    min_hours = val
                count += 1

            avg_hours = total_hours / count

            print(f"Count of issues: {count}")
            print(f"Total hours: {total_hours:.2f}")
            print(f"Average hours: {avg_hours:.2f}")
            print(f"Maximum hours: {max_hours:.2f}")
            print(f"Minimum hours: {min_hours:.2f}")
            pass




    elif choice == '5':
        print("5) ﺑﺤﺚ ﺑﺎﻟﻜﻠﻤﺔ")

        keyword = input("Enter keyword to search: ").strip().lower()
        found = False
        for r in issues_dataset:
            match = False
            for key, val in r.items():
                if keyword in str(val).lower():
                    match = True
                    break
            if match:
                print(f"ID: {r['issue_id']} | Title: {r['title']} | Priority: {r['priority']}")
                found = True

      #  if not found:
      #      print("No matching records found.")




    elif choice == '6':
        print("Exiting program. Goodbye!")
        break
    else:
        print("Invalid choice, please select from 1 to 6.")




```
