```python
scores = [82, 90, 76, 88, 95]
total = 0
count = 0
biggest = scores[0]

for n in scores:
  total += n
  count += 1
  if n > biggest:
     biggest = n

average = total / count
print(total, round(average, 2), biggest)
```
