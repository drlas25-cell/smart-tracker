```python
%%writefile tracker/functions.py
# tracker/functions.py
import os
import csv

def add_record(records, record):
    """Append one record dict to the list."""
    # TODO: أضف السجل للقائمة وارجعها
    records.append(record)
    return records

def filter_records(records, category):
    """Return records matching the given category."""
    # TODO: لفّ واشرط — أعد قائمة جديدة بالمطابقين
    return [r for r in records if r.get('category') == category]

def summarize(records, key='amount_egp'):
#def summarize(records):
    """Return count, total, average, biggest, smallest of the specified key."""
    #"""Return count, total, average, biggest, smallest of amount_egp."""
    # TODO: نمط الملخص — لو القائمة فاضية رجّع أصفار
    amounts = [float(r[key]) for r in records if key in r and r[key] is not None]
    #amounts = [float(r['amount_egp']) for r in records if 'amount_egp' in r]

    if not amounts:
        return {
            'count': 0,
            'total': 0.0,
            'average': 0.0,
            'biggest': 0.0,
            'smallest': 0.0
        }

    total_val = sum(amounts)
    count_val = len(amounts)
    return {
        'count': count_val,
        'total': total_val,
        'average': total_val / count_val,
        'biggest': max(amounts),
        'smallest': min(amounts)
    }

def save_records(records, path):
    """Write records to CSV with a header row."""
    # TODO: csv.writer — صف العناوين أول حاجة
    if not records:
        return

    headers = list(records[0].keys())
    with open(path, 'w', newline='', encoding='utf-8') as f:
        writer = csv.DictWriter(f, fieldnames=headers)
        writer.writeheader()
        writer.writerows(records)


def load_records(path):
    """Read CSV back into dicts; a missing file returns an empty list."""
    # TODO: except FileNotFoundError — لو الملف مش موجود رجّع قائمة فاضية
    # TODO: amount_egp -> float بعد القراءة
    if not os.path.exists(path):
        return []

    records = []
    try:
        with open(path, 'r', newline='', encoding='utf-8') as f:
            reader = csv.DictReader(f)
            for row in reader:
                if 'amount_egp' in row:
                    try:
                        row['amount_egp'] = float(row['amount_egp'])
                    except ValueError:
                        pass
                records.append(dict(row))
    except FileNotFoundError:
        return []
    return records

def delete_record(records, index):
    """Delete the record at index and return it."""
    if index < 0 or index >= len(records):
        return None
    return records.pop(index)


def update_record(records, index, field, value):
    """Change one field of the record at index."""
    # نتأكد أن الفهرس صحيح وموجود داخل القائمة
    if index < 0 or index >= len(records):
        return None

    # نأخذ السجل المطلوب
    record = records[index]

    # نتحقق إذا كان الحقل موجوداً بالفعل في السجل
    if field in record:
        record[field] = value
        return record

    return None
```

```python
%%writefile tracker/models.py
# tracker/models.py
import csv
import os

from tracker.functions import (add_record, filter_records, summarize,
    save_records, load_records
)

from tracker.predictor import predict_one

class Record:
    """One tracker row: the five project fields."""
    #def __init__(self, date, category, amount_egp,payment_method, quantity):
        # TODO: خزّن الحقول الخمسة على الكائن — المبلغ يتحفظ رقمًا
        #pass


    def __init__(self, issue_id, title, priority, status, hours_spent, days_open, assigned_to, platform):
        self.issue_id = int(issue_id)
        self.title = str(title)
        self.priority = str(priority)
        self.status = str(status)
        self.hours_spent = float(hours_spent)
        self.days_open = int(days_open)
        self.assigned_to = str(assigned_to)
        self.platform = str(platform)




    def to_dict(self):
        # TODO: رجّع قاموس بالحقول الخمسة — بنفس الترتيب
        #pass
        return {
            "issue_id": self.issue_id,
            "title": self.title,
            "priority": self.priority,
            "status": self.status,
            "hours_spent": self.hours_spent,
            "days_open": self.days_open,
            "assigned_to": self.assigned_to,
            "platform": self.platform
        }

    def __str__(self):
        # TODO: رجّع سطرًا منسّقًا يظهر مباشرة في الشاشة
        #pass
        return f"ID:{self.issue_id} | Title:{self.title} | Priority: {self.priority} | Status: {self.status} | Hours_spent:{self.hours_spent} | Days_open: {self.days_open} | Assigned_to: {self.assigned_to} | Platform: {self.platform}"


class Tracker:
    """A list of Record objects + every behavior the menu needs."""

    def __init__(self):
         # TODO: self.records = قائمة فاضية
        #pass
        self.records = []

    def add(self, record):
         # TODO: ضيف الكائن لقائمة السجلات
        #pass
        self.records.append(record)

    def filter_by(self, category):
        # TODO: قائمة جديدة بسجلات التصنيف المطلوب فقط
        #pass
        return [r for r in self.records if r.status == category]
       # لتصفيه السجلات حسب حاله المشكله مثلا
       # return [r for r in self.records if r.status.strip().title() == categorystrip().title()]




    def summary(self):
        # TODO: قاموس الملخص — العدد والمجموع والمتوسط والأعلى والأدنى
        #       للمبلغ الأساسي — ولو القائمة فاضية رجّع أصفار
        hours = [r.hours_spent for r in self.records]
        if not hours:
            return {
                'count': 0,
                'total': 0.0,
                'average': 0.0,
                'biggest': 0.0,
                'smallest': 0.0
            }
        total_val = sum(hours)
        count_val = len(hours)
        return {
            'count': count_val,
            'total': total_val,
            'average': total_val / count_val,
            'biggest': max(hours),
            'smallest': min(hours)
        }

    def save(self, path):
        # TODO: csv + to_dict — صف العناوين أول حاجة وبعده السجلات
        # pass
        if not self.records:
            return
        headers = ["issue_id", "title", "priority", "status", "hours_spent", "days_open", "assigned_to", "platform"]
        with open(path, 'w', newline='', encoding='utf-8') as f:
            writer = csv.DictWriter(f, fieldnames=headers)
            writer.writeheader()
            for r in self.records:

                writer.writerow(r.to_dict())




    def load(self, path):
        # TODO: اقرأ الصفوف وحوّلها لسجلات الكائن داخل المتتبّع
        #       ملف مش موجود → سيب القائمة زي ما هي
        #       المبلغ لازم يرجع رقمًا بعد القراءة


        if not os.path.exists(path):
            return
        self.records = []
        try:
            with open(path, 'r', newline='', encoding='utf-8') as f:
                reader = csv.DictReader(f)
                for row in reader:
                    record_obj = Record(
                        issue_id=row["issue_id"],
                        title=row["title"],
                        priority=row["priority"],
                        status=row["status"],
                        hours_spent=row["hours_spent"],
                        days_open=row["days_open"],
                        assigned_to=row["assigned_to"],
                        platform=row["platform"]
                    )
                    self.add(record_obj)
        except FileNotFoundError:
            pass


        # tracker/models.py — الطريقة دي تنزل جوّا كلاس المتتبّع
    def top_n(self, field, n):
        """Return the n records with the biggest value of field."""
        # TODO: لفّ على السجلات وتتبّع أكبر قيمة — ورتّبهم تنازليًا
        #pass
        sorted_records = sorted(self.records, key=lambda r: getattr(r, field), reverse=True)
        return sorted_records[:n]






    # tracker/models.py — الطريقة دي تنزل جوّا كلاس المتتبّع
    def baseline_predict(self, record):
        """Rule-based prediction - the benchmark the ML must beat."""
        # نهج يعتمد على if/else البسيطة بناءً على حقل الأولوية (Priority)
        priority = str(record.get('priority', 'Medium')).lower()

        if priority == 'high':
            return 16.0  # المشاكل العالية تحتاج وقت أطول
        elif priority == 'medium':
            return 8.0   # المشاكل المتوسطة كقيمة افتراضية
        else:
            return 4.0   # المشاكل منخفضة الأولوية




    # models.py — الطريقتين دي تنزلوا جوّا كلاس المتتبّع
    def set_model(self, model, columns):
        """يربط الموديل المدرّب بالمتتبّع."""
        self.model = model
        self.columns = columns



    def predict(self, record):
        """تنبؤ الموديل + الـ baseline لسجل جديد (dict)."""
        if self.model is None or self.columns is None:
            raise ValueError("النموذج لم يتم ربطه بعد! يرجى تشغيل set_model أولاً.")

        # التنبؤ باستخدام النموذج الذكي
        model_pred = predict_one(self.model, record, self.columns)

        # التنبؤ باستخدام الـ baseline
        base_pred = self.baseline_predict(record)

        return {
            "baseline": base_pred,
            "model": round(model_pred, 1)
        }


```

```python
%%writefile tracker/analysis.py
# tracker/functions.py
import os
import csv

"""Dataset analysis - load, clean, and summarize the project dataset."""
import pandas as pd

CSV_PATH = "data/sample_records_250.csv"


#def load_dataset(path=CSV_PATH):
    # TODO: اقرا الملف وارجع النتيجة
#    pass


#def clean_dataset(df):
    # TODO: drop_duplicates ثم اتعامل مع النواقص
    # (احذف النواقص أو عوّضها — واكتب قرارك في التقرير)
 #   pass


#def summary_by_category(df, category_col, value_col):
    # TODO: groupby + mean
#    pass



def load_dataset(path=CSV_PATH):
    """Reads the CSV file into a pandas DataFrame."""
    if not os.path.exists(path):
        print(f"Error: File not found at {path}")
        return pd.DataFrame()
    return pd.read_csv(path)




def clean_dataset(df):
    """Removes duplicates and handles missing values."""
    if df.empty:
        return df

    # إزالة الصفوف المكررة بالكامل
    df_cleaned = df.drop_duplicates()

    # التعامل مع القيم الناقصة (ملء القيم الفارغة في الحقول النصية بـ 'Unknown' والرقمية بـ المتوسط)
    for col in df_cleaned.columns:
        if df_cleaned[col].isnull().any():
            if df_cleaned[col].dtype in ['float64', 'int64']:
                df_cleaned[col] = df_cleaned[col].fillna(df_cleaned[col].mean())
            else:
                df_cleaned[col] = df_cleaned[col].fillna('Unknown')

    return df_cleaned


def summary_by_category(df, category_col, value_col):
    """Groups by category_col and calculates the mean of value_col."""
    if df.empty or category_col not in df.columns or value_col not in df.columns:
        return pd.DataFrame()

    # تجميع البيانات وحساب المتوسط لكل تصنيف
    summary_df = df.groupby(category_col)[value_col].mean().reset_index()
    return summary_df


def count_by_category(df, category_col):
    """Groups by category_col and calculates the count of occurrences."""
    if df.empty or category_col not in df.columns:
        return pd.DataFrame()

    # تجميع البيانات وحساب العدد لكل تصنيف
    count_df = df.groupby(category_col).size().reset_index(name='issue_count')
    return count_df





df_raw = load_dataset()
print(f"Raw dataset shape: {df_raw.shape}")



df_clean = clean_dataset(df_raw)
print(f"Cleaned dataset shape: {df_clean.shape}")



cleaned_file_path = "data/cleaned_sample_records_250.csv"
df_clean.to_csv(cleaned_file_path, index=False)
print(f"تم حفظ الملف بنجاح في المسار: {cleaned_file_path}")


```

```python
%%writefile tracker/predictor.py
# tracker/functions.py
import os
import csv

"""Project first ML model - L9."""
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor


import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score



TARGET = "hours_spent"       # عمود الهدف في مشروعك
CAT_COLS = ["priority", "status", "platform"]      # عموداتك الفئوية - عدّلها


def prepare_data(df):
    """يجهّز X و y: صفوف نظيفة + أعمدة فئوية مشفّرة."""
    if df.empty:
        return pd.DataFrame(), pd.Series()

    # تنظيف صفوف الهدف الناقصة
    df_clean = df.dropna(subset=[TARGET])

    # فصل المتغير المستهدف
    y = df_clean[TARGET]

    # تشفير الأعمدة الفئوية واختيار الميزات
    X = pd.get_dummies(df_clean[CAT_COLS], drop_first=True)

    return X, y


def train_model(X, y, kind="linear"):
    """يدرب ويرجّع الموديل - نوع التدريب حسب kind"""
    if kind == "linear":
        model = LinearRegression()
    elif kind == "tree":
        model = DecisionTreeRegressor(random_state=42)
    else:
        raise ValueError(f"Unknown model kind: {kind}")

    model.fit(X, y)
    return model

# تجهيز البيانات وتدريب النموذج
X, y = prepare_data(df_clean)
model = train_model(X, y, kind="linear")
print("تم تجهيز البيانات وتدريب النموذج بنجاح!")
print(f"عدد الميزات المستخدمة في التدريب: {X.shape[1]}")

import pandas as pd





def predict_one(model, record, columns):
    """تنبؤ لسجل واحد (dict) - بنفس تجهيز التدريب."""
    # 1. تحويل السجل لجدول يحتوي على صف واحد
    df_one = pd.DataFrame([record])

    # 2. تشفير الأعمدة الفئوية المتطابقة مع التدريب
    X_one = pd.get_dummies(df_one, dtype=float)

    # 3. مطابقة وتطابق الأعمدة وملء الفراغات بـ 0 لضمان صحة الإدخال للنموذج
    X_one = X_one.reindex(columns=columns, fill_value=0)

    # 4. التنبؤ وإرجاع القيمة كرقم عشري float
    prediction = float(model.predict(X_one)[0])
    return prediction



def evaluate_model(X, y, kind="linear", test_size=0.3, random_state=42):
    """تقسيم صادق + قياس على المخبّأ - يرجع مقياس الأداء للمشروع."""
    # 1. تقسيم البيانات إلى تدريب واختبار
    X_train, X_test, y_train, y_test = train_test_split(
        X, y, test_size=test_size, random_state=random_state
    )

    # 2. تدريب الموديل بناءً على النوع المختار
    model = train_model(X_train, y_train, kind=kind)

    # 3. التنبؤ على مجموعتي التدريب والاختبار
    y_pred_train = model.predict(X_train)
    y_pred_test = model.predict(X_test)

    # 4. حساب المقاييس المطلوبة
    train_mae = mean_absolute_error(y_train, y_pred_train)
    mae = mean_absolute_error(y_test, y_pred_test)
    rmse = np.sqrt(mean_squared_error(y_test, y_pred_test))
    r2 = r2_score(y_test, y_pred_test)

    return {
        "train_mae": round(train_mae, 3),
        "mae": round(mae, 3),
        "rmse": round(rmse, 3),
        "r2": round(r2, 3)
    }





```

```python
import pandas as pd
from tracker.analysis import load_dataset
from tracker.predictor import prepare_data, evaluate_model
from sklearn.model_selection import train_test_split

df = load_dataset()
X, y = prepare_data(df)

# نموذجان على نفس التقسيم
res_lin = evaluate_model(X, y, kind="linear")
res_tree = evaluate_model(X, y, kind="tree")
print("linear:", {k: round(v, 2) for k, v in res_lin.items()})
print("tree:  ", {k: round(v, 2) for k, v in res_tree.items()})

# baseline L6 — نفس روح: متوسط التدريب
X_tr, X_va, y_tr, y_va = train_test_split(
    X, y, test_size=0.3, random_state=42)
base_mae = float(abs(y_va - y_tr.mean()).mean())
print("baseline MAE:", round(base_mae, 2))

# overfitting = فرق التدريب والتحقق
for name, res in (("linear", res_lin), ("tree", res_tree)):
    print(f"{name} gap:", round(res["mae"] - res["train_mae"], 2))

best = min(res_lin["mae"], res_tree["mae"])
assert base_mae > 0 and best > 0, "المقارنة اتعملت"
print("النتيجة:", "الموديل غلب الـ baseline" if best < base_mae
      else "الbaseline كسب — قولها بصراحة في التقرير")
print("Part B: ممتاز")
```

1. ما هي النماذج المقارنة؟

النموذج المرجعي (Baseline): وهو التنبؤ البسيط المبني على قواعد ثابتة (مثلاً: إذا كانت الأهمية High يحتاج لـ 16 ساعة وهكذا). قيمته هنا هي 7.08.

النموذج الخطي (Linear Regression): وهو نموذج ذكي يتعلم الأنماط خطياً.

شجرة القرار (Decision Tree): وهو نموذج يعتمد على التفرعات الشجرية.

2. تفسير النتائج بالأرقام:

أ) النموذج الخطي (Linear Regression):
الخطأ في التدريب (Train MAE) = 7.24: يعني أن متوسط الخطأ في التنبؤ ببيانات التدريب هو حوالي 7 ساعات.
الخطأ في التحقق (Val MAE) = 5.97: وهو أقل خطأ على الإطلاق وأفضل من الـ Baseline (7.08) بفارق ممتاز جداً.
معامل التحديد (R²) = 0.22: يعني أن النموذج يستطيع تفسير 22% من الاختلافات في البيانات (وهي بداية مقبولة لبيانات بسيطة
).

ب) نموذج شجرة القرار (Decision Tree):
الخطأ في التدريب (Train MAE) = 5.82: قليل جداً لأن الشجرة قوية في حفظ البيانات التي تدربت عليها.
الخطأ في التحقق (Val MAE) = 7.0: ارتفع الخطأ هنا مقارنة بالتدريب، وهو قريب جداً من الـ Baseline.
معامل التحديد (R²) = -0.1: النتيجة بالسالب تعني أن هذا النموذج ضعيف جداً ولا يناسب البيانات الحالية.

3. ظاهرة حفظ البيانات (Overfitting):

في شجرة القرار (Decision Tree): نلاحظ أن خطأ التدريب (5.82) أقل بكثير من خطأ التحقق (7.0). هذه الفجوة تدل على حدوث Overfitting (أي أن النموذج يحفظ البيانات التدريبية ولا يستطيع التنبؤ جيداً ببيانات جديدة).


في النموذج الخطي (Linear): النموذج مستقر وقدم أداءً أفضل على بيانات التحقق (5.97) مقارنة بالتدريب، مما يجعله الخيار الأكثر أماناً واستقراراً لمشروعك.

أن النموذج الخطي هو الفائز (Best Model) لأنه حقق أقل قيمة خطأ (MAE = 5.97).

أن ذكاء الآلة نجح بالفعل في تخطي التنبؤ التقليدي البسيط (Baseline MAE = 7.08).

```python
from sklearn.model_selection import cross_val_score
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor

lin_cv = -cross_val_score(LinearRegression(), X, y, cv=3,
                          scoring="neg_mean_absolute_error")
tree_cv = -cross_val_score(DecisionTreeRegressor(max_depth=3, random_state=0),
                           X, y, cv=3, scoring="neg_mean_absolute_error")
print("linear:", round(float(lin_cv.mean()), 2))   # 5.75
print("tree:  ", round(float(tree_cv.mean()), 2))  # 18.03

assert lin_cv.mean() < tree_cv.mean(), "نفس ترتيب التقسيم الواحد"
print("C: ممتاز")
```

1. لماذا نقيّم النموذج على بيانات التحقق (Validation) وليس التدريب؟
«تقييم الموديل على نفس التدريب = كذبة»: لأن النموذج قد رأى هذه الصفوف بالفعل أثناء التدريب وحفظها. التقييم على بيانات التدريب يقيس القدرة على التذكّر والحفظ وليس القدرة على التنبؤ بسجلات جديدة لم يرها من قبل. لذلك، نعتمد دائماً على بيانات التحقق لتقييم حقيقي وصادق.
2. أهمية المعامل random_state=42:
يضمن تثبيت تقسيم البيانات في كل مرة تقومي فيها بتشغيل الكود. بدون تثبيته، ستحصلين على مجموعات بيانات تحقق مختلفة في كل مرة تشغيل، مما يجعل المقارنة بين النماذج مستحيلة لأن النتائج ستتغير عشوائياً.
3. ماذا تعني مقاييس الأداء في نتائجك؟
MAE (متوسط الخطأ المطلق): يوضح متوسط كمية الخطأ بوحدتك الأساسية (الساعات). في نموذجك الخطي، متوسط الخطأ هو 5.97 ساعة، وهو أفضل بكثير من النموذج التقليدي (7.08 ساعة).
RMSE (جذر متوسط مربعات الأخطاء): يعطي وزناً أكبر وتضخيماً للأخطاء الكبيرة جداً لتنبيهك بوجود انحرافات ضخمة. قيمته في النموذج الخطي هي 7.72.
R² (معامل التحديد): يوضح النسبة المئوية للتشتت الذي يستطيع النموذج تفسيره في البيانات:
في النموذج الخطي: قيمته 0.22 (أي 22%) وهو مقبول كبداية.
في نموذج شجرة القرار: قيمته سالبة (-0.1)، وقيمة $R^2$$R^2$ أقل من صفر تعني أن النموذج سيء جداً وأدائه أسوأ من مجرد التخمين العشوائي بمتوسط الساعات!
4. كيف نكشف الـ Overfitting (حفظ البيانات) بالأرقام من نتائجك؟
في نموذج شجرة القرار (Decision Tree):
الخطأ في التدريب (Train MAE) = 5.82
الخطأ في التحقق (Val MAE) = 7.0
الفجوة (Gap) = 1.18 (خطأ التدريب أقل بكثير من التحقق، مما يعني أن النموذج يحفظ ولا يفهم).
في النموذج الخطي (Linear Regression):
الخطأ مستقر بل وأفضل في التحقق (5.97) مقارنة بالتدريب (7.24)، مما يعني عدم وجود Overfitting وأن النموذج مستقر وقابل للتعميم.
